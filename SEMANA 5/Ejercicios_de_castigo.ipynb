```python
while True:
    try:
        numero1 = int(input("Ingresa el primer número entero: "))
        numero2 = int(input("Ingresa el segundo número entero: "))
        break

    except ValueError:
        print(
            "Error: Debes ingresar números enteros. "
            "Inténtalo nuevamente."
        )

try:
    resultado = numero1 / numero2

    print(f"{numero1} dividido entre {numero2} es: {resultado}")

except ZeroDivisionError:
    print(
        "Error: No es posible dividir entre cero. "
        "El segundo número debe ser diferente de 0."
    )
```

```python
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)


frase = input("Ingresa una frase que contenga la palabra Alebrije: ")

print("Longitud de la frase:", len(frase))
print("Frase en mayúsculas:", frase.upper())

frase_reemplazada = frase.replace("Alebrije", "maravilla")

print("Frase reemplazada:", frase_reemplazada)
print("Cantidad de palabras:", contar_palabras(frase))
```

```python
votos = {
    "Rojo": 0,
    "Azul": 0,
    "Verde": 0
}

cantidad_votos = 0

while cantidad_votos < 5:
    color = input("Ingresa tu voto (Rojo, Azul o Verde): ").capitalize()

    if color in votos:
        votos[color] += 1
        cantidad_votos += 1
    else:
        print("Color inválido. Inténtalo nuevamente.")

print("\nResultados:")

for color, total in votos.items():
    print(f"{color}: {total} votos")

mayor_cantidad = max(votos.values())

ganadores = []

for color, total in votos.items():
    if total == mayor_cantidad:
        ganadores.append(color)

if len(ganadores) == 1:
    print(
        f"El color ganador es {ganadores[0]} "
        f"con {mayor_cantidad} votos."
    )
else:
    print(
        f"Hay un empate entre {', '.join(ganadores)} "
        f"con {mayor_cantidad} votos cada uno."
    )
```
