```python
def sumar_tupla(tupla):
    total = 0
    for numero in tupla:
        total += numero
    return total

def seccion_tuplas( ):
    print("\nSección de tuplas")
    numeros = (50,100,150,200,250)
    print("Tupla de números:", numeros)
    print("Tercer elemento:", numeros[2])
    
    try:
        e1 = float(input("Ingresa el primer numero adicional: "))
        e2 = float(input("Ingresa el segundo numero adicional: "))
    except ValueError:
        print("Error: Debe ingresar un numero válido.")
        return  
    nueva_tupla = numeros + (e1, e2)
    print("Nueva tupla:", nueva_tupla)

    lista_ordenada = list(nueva_tupla)
    lista_ordenada.sort()
    print("Lista ordenada:", lista_ordenada)
    print("Suma de los elementos de la tupla:", sumar_tupla(nueva_tupla))

def buscar_telefono(contactos, nombre):
    return contactos.get(nombre)

def seccion_diccionario():
    print("\nSección de diccionarios")
    contactos = {
        "Sofia Gutierrez": "5612073746",
        "Gloria Luna": "5971284753",
        "Rosa Esther": "4421704068"
    }
    nombre_nuevo = input("Ingresa el nombre del nuevo contacto: ")
    telefono_nuevo = input("Ingresa el número de teléfono del nuevo contacto: ")
    contactos[nombre_nuevo] = telefono_nuevo
    print("\nNombres de los contactos registrados:")
    for nombre in contactos.keys():
        print(nombre)
    nombre_buscado = input("\nIngresa el nombre del contacto a buscar: ")
    telefono_encontrado = buscar_telefono(contactos, nombre_buscado)   
    if telefono_encontrado is not None:
        print(f"El número de teléfono de {nombre_buscado} es: {telefono_encontrado}")
    else:
        print(f"No se encontró el contacto con el nombre {nombre_buscado}.")

def seccion_excepciones():
    print("\nSección de excepciones")
    try:
        n1 = int(input("Ingresa el primer numero entero: "))
        n2 = int(input("Ingresa el segundo numero entero: "))
        print("suma:", n1 + n2)
        print("division:", n1/n2)
    except ValueError:
        print("Error: Debes ingresar un número entero valido.")
    except ZeroDivisionError:
        print("Error: No se puede dividir entre cero, escoge otro numero.")

def seccion_strings():
    print("\nSección de strings")
    contar_palabras = lambda texto: len(texto.split())
    texto = input("Ingresa un texto: ")
    print("Texto ingresado:", texto)
    print("Longitud del texto:", len(texto))
    print("Texto en mayúsculas:", texto.upper())
    print("Cantidad de palabras:", contar_palabras(texto))

def mostrar_menu():
    print("\nMenú de opciones:")
    print("1. Sección de tuplas")
    print("2. Sección de diccionarios")
    print("3. Sección de excepciones")
    print("4. Sección de strings")
    print("5. Salir")
    opcion = input("Seleccione una opción (1-5): ")
    if opcion == "1":
        seccion_tuplas()
    elif opcion == "2":
        seccion_diccionario()
    elif opcion == "3":
        seccion_excepciones()
    elif opcion == "4":
        seccion_strings()
    elif opcion == "5":
        print("Saliendo del programa.")
        return False
    else:
        print("Opción inválida. Por favor, seleccione una opción válida.")   

mostrar_menu()
```
