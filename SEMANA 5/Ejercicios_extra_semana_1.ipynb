```python
def sumar_calificaciones(calificaciones):
    total = 0

    for calificacion in calificaciones:
        total += calificacion

    return total


calificaciones = (6.5, 8.0, 7.5, 9.5, 10.0)

print("Tercera calificación:", calificaciones[2])

nueva_calificacion1 = float(input("Nueva calificación 1: "))
nueva_calificacion2 = float(input("Nueva calificación 2: "))

nueva_tupla = calificaciones + (
    nueva_calificacion1,
    nueva_calificacion2
)

print("Nueva tupla:", nueva_tupla)

lista_ordenada = list(nueva_tupla)
lista_ordenada.sort(reverse=True)

print("Lista ordenada (mayor a menor):", lista_ordenada)

suma_total = sumar_calificaciones(nueva_tupla)

print("Suma total:", suma_total)
```

```python
def buscar_telefono(agenda, nombre):
    return agenda.get(nombre)


agenda = {
    "Jose": "5612054122",
    "Luis": "5523388142",
    "Esteban": "5555955624"
}

nombre_nuevo = input("Nombre del nuevo contacto: ")
telefono_nuevo = input("Teléfono del nuevo contacto: ")

agenda[nombre_nuevo] = telefono_nuevo

print("Contactos registrados:", ", ".join(agenda.keys()))

nombre_buscado = input("Nombre a buscar: ")

telefono = buscar_telefono(agenda, nombre_buscado)

if telefono is not None:
    print(f"El teléfono de {nombre_buscado} es: {telefono}")
else:
    print("Contacto no encontrado.")
```

```python
try:
    numero1 = int(input("Primer número entero: "))
    numero2 = int(input("Segundo número entero: "))

    resultado = numero1 / numero2

    print(f"{numero1} dividido entre {numero2} es: {resultado}")

except ValueError:
    print("Error: Debes ingresar números enteros.")

except ZeroDivisionError:
    print(
        "Error: No es posible dividir entre cero. "
        "Ingresa un divisor distinto de 0."
    )
```

```python
def contar_palabras(texto):
    palabras = texto.split()
    return len(palabras)


mensaje = "Bad bunny es el mejor cantante de todos los tiempos"

print("Longitud del mensaje:", len(mensaje))
print("En mayúsculas:", mensaje.upper())

texto_reemplazado = mensaje.replace("Bad bunny", "El conejo malo")
print("Texto reemplazado:", texto_reemplazado)

total_palabras = contar_palabras(mensaje)
print("Palabras totales:", total_palabras)
```
