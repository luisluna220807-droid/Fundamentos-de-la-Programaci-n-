```python
numero = int(input("ingrese un numero entero positivo: "))

numero_original = numero
suma = 0

while numero > 0:
    digito = numero % 10
    suma = suma + digito
    numero = numero // 10

print("la suma de los digitos de ", numero_original, "es", suma)
```

```python
num = int(input("ingrese el numero que desea dividir: "))
num_original = num
divisor = int(input("ingrese el el numero por el cual desea dividir: "))
cociente = 0

while num >= divisor:
    num = num - divisor
    cociente = cociente + 1
    residuo = num

print("el resultado de dividir", num_original, "entre", divisor, "es", cociente, "con un residuo de", residuo)
```

```python
n = int(input("ingresa un numero entero positivo: "))

suma = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        suma -= i
    else:
        suma += i

print("el resultado de la serie es", suma)
```

```python
n = int(input("ingresa un numero entero positivo: "))

suma = 0.0

for i in range(1, n + 1):
    suma += 1 / i

print("la suma de la serie es :", format(suma, ".2f"))
```
