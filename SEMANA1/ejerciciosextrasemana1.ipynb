# Ejercicio 1 

```python
total_de_la_cuenta = float(input("Ingrese el total de la cuenta: "))
porcentaje_de_propina = float(input("Ingrese el porcentaje de propina que desea dejar: "))
numero_de_personas_a_dividir_la_cuenta = float(input("Ingrese el numero de personas a dividir la cuenta: "))
propina = total_de_la_cuenta * (porcentaje_de_propina / 100)
total_de_la_cuenta_con_propina = total_de_la_cuenta + propina
pago_por_persona = total_de_la_cuenta_con_propina / numero_de_personas_a_dividir_la_cuenta
print(f"El total de la cuenta es: ${total_de_la_cuenta}")
print(f"El porcentaje de propina es: {porcentaje_de_propina}%")
print(f"EL numero de personas a dividir la cuenta es: {numero_de_personas_a_dividir_la_cuenta}")
print(f"La propina es de: ${propina}")
print(f"El total de la cuenta con propina es de: ${total_de_la_cuenta_con_propina}")
print(f"El pago por persona es de: ${pago_por_persona}")
```

 # Ejercicio 3
 

```python
calificacion_primer_parcial = int(input("Ingrese su calificacion del primer parcial: "))
calificacion_segundo_parcial = int(input("Ingrese su calificacion del segundo parcial: "))
calificacion_tercer_parcial = int(input("Ingrese su calificacion del tercer parcial: "))
calificacion_final = ( 0.3 * calificacion_primer_parcial + 0.3 * calificacion_segundo_parcial + 0.4 * calificacion_tercer_parcial)
print("Su calificacion del primer parcial es: ", calificacion_primer_parcial)
print("Su calificacion del segundo parcial es: ", calificacion_segundo_parcial)
print("Su calificacion del tercer parcial es: ", calificacion_tercer_parcial)
print("Su calificacion final es: ", calificacion_final)
```
