#  Tabla de pitagoras 

```python
tabla = []

for fila in range(1, 21):
    fila_actual = []
    for columna in range(1, 21):
        resultado = fila * columna
        fila_actual.append(resultado)
    tabla.append(fila_actual)

def imprimir_tabla(tabla):
    for fila_actual in tabla:
        for n in fila_actual:
            print(n, end="\t")
        print()

def consultar (tabla, fila, columna):
    return tabla[fila - 1][columna - 1]

print("Tabla de Pitagoras")
print()
imprimir_tabla(tabla)
print()

fila = int(input("Ingresa el numero de fila (1-20): "))
columna = int(input("Ingresa el numero de columna (1-20): "))

if 1 <= fila <= 20 and 1 <= columna <= 20:
    resultado = consultar (tabla, fila, columna)
    print(f"el resultado de {fila} x {columna} es: {resultado}")
else:
    print("Error: los valores deben de estar entre 1 y 20.")
```
