```python
def sumar_lista(lista):
    total = 0
    for elemento in lista:
        total += elemento
    return total

numeros = [10, 20, 30, 40]
print(f"La suma de los elementos es: {sumar_lista(numeros)}")
```

```python
def maximo_lista(lista):
    mayor = lista[0]
    for elemento in lista:
        if elemento > mayor:
            mayor = elemento
    return mayor

numeros = [7, 3, 9, 2]
print(f"El elemento máximo es: {maximo_lista(numeros)}")
```

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
suma_diagonal = 0
for i in range(len(matriz)):
    suma_diagonal += matriz[i][i]
print(f"La suma de la diagonal principal es: {suma_diagonal}")
```

```python
def mostrar_promedio(calificaciones):
    prom = sum(calificaciones) / len(calificaciones)
    print(f"El promedio es: {prom:.2f}")

mostrar_promedio([8, 9, 10, 7])
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

num = 5
print(f"El factorial de {num} es: {factorial(num)}")
```

```python
def procesar_tupla(t):
    tercero = t[2]
    suma_total = sum(t)
    print(f"El tercer elemento es: {tercero}")
    print(f"La suma de la tupla es: {suma_total}")

procesar_tupla((4, 8, 15, 16, 23, 42))
```
