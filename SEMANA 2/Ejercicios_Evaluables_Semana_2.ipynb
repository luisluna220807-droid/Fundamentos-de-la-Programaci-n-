 # Actividad Evaluable 2: Cobro de Museo 

```python
print("==========================================")
print("   MUSEO DE ANTROPOLOGÍA E HISTORIA")
print("==========================================")

tabla_descuentos = {
    "Adulto mayor": 0.12,
    "Profesor": 0.10,
    "Estudiante": 0.10,
    "Ninguno": 0.00
}

while True:
    try:
        cantidad = int(input("\nIngrese el número de visitantes: "))

        if cantidad <= 0
            print("Error: debe ingresar un número mayor que 0.")
            continue

        break

    except ValueError:
        print("Error: ingrese solamente un número entero.")

total_general = 0
total_descuentos = 0

for visitante in range(1, cantidad + 1):

    print("\n==========================================")
    print("VISITANTE", visitante)
    print("==========================================")

    while True:
        try:
            edad = int(input("Ingrese la edad: "))

            if edad < 0:
                print("Error: la edad no puede ser negativa.")
                continue

            break

        except ValueError:
            print("Error: ingrese una edad válida.")

    if edad < 3:

        precio_base = 0
        categoria = "Menor de 3 años"

    elif edad <= 17:

        precio_base = 60
        categoria = "Menor de edad"

    else:

        precio_base = 80
        categoria = "Mayor de 18 años"

    if edad < 3:

        tipo = "Menor de 3 años"
        descuento = tabla_descuentos["Ninguno"]

        monto_descuento = 0
        precio_final = 0

        print("\nCategoría:", categoria)
        print("Tipo de visitante:", tipo)
        print("Entrada: GRATIS")
        print("Total a pagar: $0.00")

        continue

    elif edad < 22:

        tipo = "Estudiante"
        descuento = tabla_descuentos["Estudiante"]

    elif edad > 60:

        tipo = "Adulto mayor"
        descuento = tabla_descuentos["Adulto mayor"]

    else:

        while True:

            print("\nTipo de visitante:")
            print("1. Profesor")
            print("2. Estudiante")

            opcion = input("Seleccione una opción: ")

            if opcion == "1":

                tipo = "Profesor"
                descuento = tabla_descuentos["Profesor"]

                break

            elif opcion == "2":

                tipo = "Estudiante"
                descuento = tabla_descuentos["Estudiante"]

                break

            else:

                print("Opción inválida.")
                print("Seleccione 1 para Profesor o 2 para Estudiante.")

                continue

    monto_descuento = precio_base * descuento
    precio_final = precio_base - monto_descuento
    total_general += precio_final
    total_descuentos += monto_descuento

    print("\n---------- RESUMEN ----------")
    print("Edad:", edad)
    print("Categoría:", categoria)
    print("Tipo de visitante:", tipo)
    print(f"Precio base: ${precio_base:.2f}")
    print(f"Descuento aplicado: {descuento * 100:.0f}%")
    print(f"Monto descontado: ${monto_descuento:.2f}")
    print(f"Total a pagar: ${precio_final:.2f}")

print("\n==========================================")
print("           TOTAL DETALLADO")
print("==========================================")

print("Número de visitantes:", cantidad)
print(f"Total de descuentos: ${total_descuentos:.2f}")
print(f"TOTAL A PAGAR: ${total_general:.2f}")

print("==========================================")
print("      GRACIAS POR VISITAR EL MUSEO")
print("==========================================")
```
