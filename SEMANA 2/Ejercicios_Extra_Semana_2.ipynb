```python
total = 0
visitante = 1

while True:
    costo = int(input(f"Costo del boleto {visitante}: "))

    if costo == 0:
        print("Menor de 3 años, sin cargo")
        visitante += 1
        continue

    total += costo

    if total >= 100:
        break

    visitante += 1

print(f"Total acumulado: ${total}")
```

```python
num_visitantes = int(input("Número de visitantes: "))

adultos = 0
suma_edades = 0

for i in range(1, num_visitantes + 1):
    edad = int(input(f"Edad del visitante {i}: "))
    suma_edades += edad

    if edad >= 18:
        adultos += 1

promedio = suma_edades / num_visitantes

print(f"Adultos: {adultos}")
print(f"Promedio de edad: {promedio:.2f}")
```

```python
precio = 45
descuento = 0.12

total = precio - (precio * descuento)

print(f"Total: ${total:.2f}")
```
